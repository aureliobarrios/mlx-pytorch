## Imports

```python
import torch
import mlx.core as mx
```

```python
#defining a network module
import mlx.nn as mx_nn

class MLXNet(mx_nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = mx_nn.Linear(10, 2)
    
    def __call__(self, x):
        return self.fc(x)
```

```python
import mlx.optimizers as mx_opt

#initialize our mlx model and optimizer
model = MLXNet()
optimizer = mx_opt.Adam(learning_rate=1e-3)
```

```python
#define a loss function
def loss_fn(model, x, y):
    prediction = model(x)
    return mx_nn.losses.mse_loss(prediction, y)
```

```python
#create a functional transformation using value and gradient
loss_and_grad_fn = mx.value_and_grad(loss_fn)
```

```python
#dummy data for training
X = mx.random.normal([32, 10])
y = mx.random.normal([32, 2])
```

```python
#training exectution loop
for epoch in range(10):
    #forward and backward passes occur together
    loss, grads = loss_and_grad_fn(model, X, y)
    #update the model parameters using computed gradients
    optimizer.update(model, grads)
    #force evaluation of graph computations to output loss value
    mx.eval(model.parameters(), optimizer.state)

    print(f"Epoch {epoch}: Loss = {loss.item():.4f}")
```
